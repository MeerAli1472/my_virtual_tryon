```python
import gradio as gr
import os
from PIL import Image as PILImage
import cv2

# Set paths
input_dir = '/content/drive/MyDrive/TryYours-Virtual-Try-On-main/static'
cloth_path = os.path.join(input_dir, 'cloth_web.jpg')
origin_path = os.path.join(input_dir, 'origin_web.jpg')
result_path = os.path.join(input_dir, 'finalimg.png')

# Create input directory if not exists
os.makedirs(input_dir, exist_ok=True)

# Function to save uploaded images
def save_and_resize(image, save_path, size=(768,1024)):
    img = PILImage.fromarray(image)
    img = img.resize(size, PILImage.LANCZOS)
    img.save(save_path)

# Function to display input images side by side and result after Try On
def try_on(person_image, garment_image):
    if person_image is None or garment_image is None:
        return None, None, "Please upload both images first!"

    # Save uploaded images
    save_and_resize(person_image, origin_path)
    save_and_resize(garment_image, cloth_path)

    # Run your try-on script
    os.system(f'python /content/drive/MyDrive/TryYours-Virtual-Try-On-main/main2.py')

    # Read images back for display
    person_img = cv2.cvtColor(cv2.imread(origin_path), cv2.COLOR_BGR2RGB)
    cloth_img = cv2.cvtColor(cv2.imread(cloth_path), cv2.COLOR_BGR2RGB)

    if os.path.exists(result_path):
        result_img = cv2.cvtColor(cv2.imread(result_path), cv2.COLOR_BGR2RGB)
    else:
        result_img = None

    return person_img, cloth_img, result_img

# Build Gradio interface
with gr.Blocks() as demo:
    gr.Markdown("## 🧥👤 Virtual Try-On System")

    with gr.Row():
        person_image_input = gr.Image(label="Upload Person Image")
        garment_image_input = gr.Image(label="Upload Garment Image")

    tryon_button = gr.Button("Try On!")

    with gr.Row():
        person_output = gr.Image(label="Person Image")
        garment_output = gr.Image(label="Garment Image")

    result_output = gr.Image(label="Try-On Result")

    tryon_button.click(
        try_on,
        inputs=[person_image_input, garment_image_input],
        outputs=[person_output, garment_output, result_output]
    )

# Launch the Gradio app inside Colab
demo.launch(share=True)  # set share=True if you want public link too
```
